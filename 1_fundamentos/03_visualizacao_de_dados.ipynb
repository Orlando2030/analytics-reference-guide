#### 4.2. Heatmap de correlações

```python
# Exemplo com dados de reviews (nota, tempo de entrega etc.)
merged = pedidos.merge(avaliacoes_pedidos, on="id_pedido", how="inner")
merged['nota_avaliacao'] = merged['nota_avaliacao'].astype(float)

correlacoes = merged[['tempo_entrega', 'nota_avaliacao']].corr()
print(correlacoes)
sns.heatmap(correlacoes, annot=True, linewidths= 2, linecolor='orange').set(title="Correlação entre tempo de entrega e avaliação", xlabel="Tempo de Entrega (dias)", ylabel="Nota de Avaliação" )
```

Interpretação:

•	Existe uma correlação negativa moderada entre o tempo de entrega e a nota de avaliação.

•	Ou seja, quanto maior o tempo de entrega, menor tende a ser a nota dada pelo cliente.

•	Isso reforça a importância da logística no nível de satisfação do cliente.
