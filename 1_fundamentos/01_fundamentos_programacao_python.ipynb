Uma variável é como uma “caixa” na memória do computador usada para guardar um valor. 

Em Python, você pode criar uma variável usando o sinal de igual (=) e atribuindo um valor a ela. Por exemplo:

```python
# Variáveis e tipos básicos
nome = "Orlando"    # str: cadeia de caracteres (texto)
idade = 41          # int: números inteiros
altura = 1.80       # float: números decimais
usa_oculos = True   # bool: valores booleanos (True ou False)
```

### Manipulação de Strings
Manipulação de strings é o conjunto de técnicas e operações que usamos para trabalhar com textos em programação.


	•	Declaração de strings ('texto', "texto", """texto""")
	•	Métodos de strings:
	•	.lower(), .upper(), .capitalize()
	•	.strip(), .lstrip(), .rstrip()
	•	.replace(), .split(), .join()
	•	.find() e .count()
	•	Formatação:
	•	f"Olá {nome}!"
	•	"{}".format(valor)
	•	Fatiamento (slicing) de strings
	•	Checagens: .isdigit(), .isalpha(), .startswith(), .endswith()


```python
texto = " Python é incrível! "
print(texto.strip())       # remove espaços
print(texto.upper())       # maiúsculas
print(texto.lower())       # minúsculas
print(texto.replace("incrível", "fantástico"))  
print(texto.split())       # quebra por espaço
print(texto.find('y'))   # posição do primeiro 'y'
print(texto.count('n'))  # conta quantos 'n' existem
print(texto.isnumeric()) # se é numérico
print(texto.isalpha())   # se é alfabético
print(len(texto))        # comprimento
```

### Operadores
Operadores são símbolos especiais usados em programação para realizar operações sobre valores e variáveis.

Eles servem para calcular, comparar, combinar ou manipular dados.

Pensa neles como os sinais da matemática (+, -, =) mas com muito mais funções.

```python
# Operadores matemáticos
a = input("Digite o Primeiro Numero:")
b = input("Digite o Segundo Numero:")

# Exibir os números digitados
print('--------------------------------')
print(f'O Primeiro numero digitado é:{a}')
print(f'O Segundo numero digitado é:{b}')
print('--------------------------------')

# Converter para inteiros
a_t = int(a)
b_t = int(b)

# Exibir calculos
print('--------------------------------')
print("Soma:", a_t + b_t)
print("Subtração:", a_t - b_t)
print("Multiplicação:", a_t * b_t)
print("Divisão:", a_t / b_t)
print("Divisão inteira:", a_t // b_t)
print("Resto:", a_t % b_t)
print("Exponenciação:", a_t ** b_t)
print('--------------------------------')
```

```python
# Exemplos de operadores lógicos em Python

x = True
y = False

# AND (E lógico) - True somente se ambos forem True
print("AND:", x and y)  # False, porque y é False

# OR (OU lógico) - True se pelo menos um for True
print("OR:", x or y)    # True, porque x é True

# NOT (NÃO lógico) - Inverte o valor
print("NOT x:", not x)  # False, porque x é True
print("NOT y:", not y)  # True, porque y é False
```

Exercício: criar programa que calcula IMC(Atualmente o Índice de Massa Corporal) e mostra classificação.
IMC =  Peso / Altura²
IMC < 18,5: Abaixo do peso
18,5 <= IMC < 25: Peso normal
25 <= IMC < 30: Sobrepeso
30 <= IMC < 40: Obesidade
IMC >= 40: Obesidade mórbida

```python
# Incercao de dados Altura e Peso
altura = input("Digite sua altura em metros: ")
peso = input("Digite seu peso em Kg: ")

# Conversão de string para float
a = float(altura)
p = float(peso)     

# Cálculo do IMC
imc = p / (a ** 2)

# Exibição do resultado com diferentes formatações
print("IMC:", imc) # Todas casas decimais
print(f"IMC: {imc:.2f}") # 2 casas decimais
print(f"IMC: {imc:.3f}") # 3 casas decimais         
```

### Condições
Condições são instruções que permitem ao programa tomar decisões.

Elas verificam se uma expressão lógica é verdadeira (True) ou falsa (False) e, com base nisso, escolhem qual parte do código deve ser executada.

```python
# Verficar se o aluno está aprovado, em recuperação ou reprovado.
nota = input("Digite a nota do aluno: ")

# Convertendo a nota para float 
nota = float(nota)

# Verificando a situação do aluno
if nota >= 0 and nota <= 10:
    if nota >= 7:
        print(f"Sua nota é {nota} e você está aprovado.")
    elif nota >= 5:
        print(f"Sua nota é {nota} e você está em recuperação.")
        print("Recuperação")
    else:
        print(f"Sua nota é {nota} e você está reprovado.")
        print("Reprovado")
else:
    print("Nota inválida. Digite uma nota entre 0 e 10.")
```

```python
# Verificar se é maior ou menor de idade
idade = input("Digite sua idade: ")

# Convertendo a idade para inteiro
idade = int(idade)

# Verificando se é maior ou menor de idade
if idade >= 0:
    if idade >= 18:
        print("Sua idade é {idade} e você é maior de idade.")
    else:
        print("Sua idade é {idade} e você é menor de idade.")
else:
    print("Idade inválida.")
    print("A idade deve ser um número inteiro positivo.")
```

### Loops
Loops são estruturas de repetição que permitem executar um bloco de código várias vezes, de forma automática, até que uma condição seja satisfeita.

Em vez de repetir o mesmo código várias vezes manualmente, você usa um loop para fazer isso por você.

```python
# For loop
# for → inicia um loop, que vai repetir um bloco de código várias vezes.
# i → é a variável de controle do loop. Ela vai receber cada valor do intervalo.
# range(1, 11) → gera uma sequência de números de 1 até 10 (o 11 não é incluído).
# : → indica que o próximo bloco de código pertence ao loop.
for i in range(1, 11):
# i vai assumir os valores: 1, 2, 3, …, 10. Cada valor será usado dentro do loop.
    print(f"Tabuada do 5 x {i} = {5*i}")
```

```python
# While loop
contador = 1         # Inicializa o contador com 0
while contador < 11:   # Enquanto o contador for menor que 11
    print(f"Tabuada do 7 x {contador} = {7*contador}")
    contador += 1     # Adiciona 1 ao contador a cada iteração
```

### Dicionários, Tuplas, Sets
São três estruturas de dados muito importantes do Python, cada uma com suas características.

#### Dicionarios
Estrutura que armazena pares de chave e valor.

Pensa como um “mini banco de dados” ou um dicionário da vida real: você procura uma palavra (chave) e encontra o significado (valor).

```python
# São usados quando precisamos guardar informações em formato chave → valor (como se fosse uma ficha de cadastro).

# Dicionário
pessoa = {"nome": "Ana", "idade": 25, "cidade": "São Paulo"}
print(pessoa["nome"])
print(pessoa["idade"])
print(pessoa["cidade"]) 
```

#### Tuplas
Estrutura ordenada de itens, imutável (não pode ser alterada depois de criada).

	•Parecida com listas, mas não podem ser modificadas.

	•Útil quando você precisa garantir que os dados não mudem.
	
	•Também são mais rápidas e consomem menos memória que listas.

```python
# São listas imutáveis (não podem ser alteradas).
#Ótimas para representar valores fixos que não mudam.

# Tupla
coordenadas = (10.0, 20.0)
print(coordenadas[0])
```

#### Set
Estrutura que guarda uma coleção de valores únicos e sem ordem.

Pensa em um conjunto matemático.

	•Não permite valores repetidos.

	•Não garante ordem dos elementos.
    
	•Útil para remover duplicatas ou fazer operações de conjuntos (união, interseção, diferença).

```python
# São coleções sem ordem e sem elementos repetidos.
# Muito úteis quando queremos eliminar duplicatas.

# Set
numeros = {1, 2, 2, 3, 4}
print(numeros)
```

## Funções
Uma função é um bloco de código reutilizável que executa uma tarefa específica.

Em vez de repetir o mesmo código várias vezes, você cria a função uma vez e depois pode chamá-la sempre que precisar.

```python
# Função simples

def saudacao():
    print("Olá, bem-vindo ao Python!")

saudacao()  # Chamando a função
```

```python
# Função com parâmetros
def media(lista):
    return sum(lista) / len(lista)

notas = [8, 7.5, 9, 6]
print("Média:", media(notas))
```

```python
def soma(a, b):
    return a + b

print(soma(5, 3))
```

### Arquivos CSV

```python
import csv

# Escrever CSV
with open("alunos.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Nome", "Nota"])
    writer.writerow(["Ana", 8])
    writer.writerow(["Carlos", 7])

# Ler CSV
with open("alunos.csv", "r") as f:
    reader = csv.reader(f)
    for linha in reader:
        print(linha)
```

### Boas práticas (PEP8) e mini-projeto

```python
# Mini-projeto:
# 1. Recebe nomes e notas de alunos
# 2. Salva em CSV
# 3. Lê CSV e calcula média
import csv

alunos = []
for i in range(3):
    nome = input("Nome: ")
    nota = float(input("Nota: "))
    alunos.append([nome, nota])

with open("alunos_final.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Nome", "Nota"])
    writer.writerows(alunos)

# Ler e calcular média
notas = []
with open("alunos_final.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        notas.append(float(row["Nota"]))
print("Média da turma:", sum(notas)/len(notas))
```
