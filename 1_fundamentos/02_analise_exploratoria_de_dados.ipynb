# 2. Limpeza e Tratamento dos Dados
	•	Remover colunas irrelevantes
	•	Tratar valores ausentes
	•	Converter datas (pd.to_datetime)
	•	Criar colunas úteis: tempo de entrega, tempo até avaliação, tempo de envio etc.

#### 2.1 Pedidos

```python
# Tabela de pedidos
pedidos.head()
```

```python
# Exibindo informações sobre o DataFrame de pedidos
pedidos.info()
```

```python
pedidos = pedidos.rename(columns={
    'order_id': 'id_pedido',
    'customer_id': 'id_cliente',
    'order_status': 'status_pedido',
    'order_purchase_timestamp': 'data_compra',
    'order_approved_at': 'data_aprovacao',
    'order_delivered_carrier_date': 'data_envio_transportadora',
    'order_delivered_customer_date': 'data_entrega_cliente',
    'order_estimated_delivery_date': 'data_entrega_estimativa'
})

pedidos.head()
```

```python
# Exibindo a frequência de status dos pedidos
pedidos['status_pedido'].value_counts()
```

```python
# Traduzindo os status dos pedidos para o português
pedidos['status_pedido'] = pedidos['status_pedido'].replace({
    'delivered': 'entregue',
    'shipped': 'enviado',
    'canceled': 'cancelado',
    'unavailable': 'indisponível',
    'invoiced': 'faturado',
    'processing': 'processando',
    'created': 'criado',
    'approved': 'aprovado'
})

pedidos['status_pedido'].value_counts()
```

```python
# Exibindo informações sobre o DataFrame de clientes
clientes.info()
```

```python
clientes = clientes.rename(columns={
    'customer_id': 'id_cliente',
    'customer_unique_id': 'id_unico_cliente',
    'customer_zip_code_prefix': 'cep_prefixo',
    'customer_city': 'cidade_cliente',
    'customer_state': 'estado_cliente'
})
clientes.info()
```

```python
clientes.head()
```

```python
itens_pedidos.head()
```

# 3. Etapas para Realizar as Junções dos Datasets 
#### (orders, customers, order_items, order_reviews, products)

####  3.1. Juntar orders (Pedidos) + customers (Clientes)

```python
# Endendo o DataFrame de clientes
print(clientes.shape)
print(clientes.columns)
```

```python
df = pedidos.merge(clientes, on='id_cliente', how='left')
df.head()
```

```python
print(df.shape)
print(df.columns)
```

#### 3.2. Juntar com order_items (Itens dos Pedidos)

```python
print(itens_pedidos.shape)
print(itens_pedidos.columns)
```

```python
df = df.merge(itens_pedidos, on='id_pedido', how='left')
df.head()
```

#### 3.3. Juntar com order_reviews (Avaliação dos Pedidos)

```python
print(avaliacoes_pedidos.shape)
print(avaliacoes_pedidos.columns)
```

```python
df = df.merge(avaliacoes_pedidos, on='id_pedido', how='left')
df.head()
```

#### 3.4. Juntar com products (Produtos)

```python
print(produtos.shape)
print(produtos.columns)
```

```python
df =df.merge(produtos, on='id_produto', how='left')
df.head()
```

```python
print(df.shape)
print(df.columns)
```

#### 3.5. Criando Base Completa

```python
df.to_csv("base_vendas_completa.csv", index=False)
```
